### Q.2 Questions on Basic NumPy Array:
a) Reverse the NumPy array: arr = np.array([1, 2, 3, 6, 4, 5])

b) Find the most frequent value and their indice(s) in the following arrays:
i. x = np.array([1,2,3,4,5,1,2,1,1,1])
ii. y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])

```python
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print(arr)
print(reversed_arr)

from collections import Counter

def most_frequent_value_and_indices(arr):
    counts = Counter(arr)
    if not counts:
        return "Array is empty", []
    most_common_value = counts.most_common(1)[0][0]
    indices = np.where(arr == most_common_value)[0]
    return most_common_value, indices

x = np.array([1,2,3,4,5,1,2,1,1,1])
most_freq_x, indices_x = most_frequent_value_and_indices(x)
print(f"\nArray x: {x}")
print(f"Most frequent value in x: {most_freq_x}")
print(f"Indices of most frequent value in x: {list(indices_x)}")

y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ])
most_freq_y, indices_y = most_frequent_value_and_indices(y)
print(f"\nArray y: {y}")
print(f"Most frequent value in y: {most_freq_y}")
print(f"Indices of most frequent value in y: {list(indices_y)}")
```

### Q.4 Write program to create an 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100 using linspace(). Print the dimensions of the array, shape, total elements, the data type of each element and total number of bytes consumed by the array. Find the transpose of this array using reshape() attribute. Can we do the same with T attribute?

```python
vaani_array = np.linspace(10, 100, 25)
print(vaani_array)

print(f"\nDimensions: {vaani_array.ndim}")
print(f"Shape: {vaani_array.shape}")
print(f"Total elements: {vaani_array.size}")
print(f"Data type: {vaani_array.dtype}")
print(f"Total number of bytes: {vaani_array.nbytes} bytes")

reshaped_for_transpose = vaani_array.reshape(-1, 1)
transposed_by_reshape = reshaped_for_transpose.T
print(f"\nOriginal 1D array reshaped to a column vector (shape {reshaped_for_transpose.shape}):\n{reshaped_for_transpose}")
print(f"Transpose using reshape() and .T (shape {transposed_by_reshape.shape}):\n{transposed_by_reshape}")

transposed_by_T_1d = vaani_array.T
print(f"\nTranspose of 1D array using .T (shape {transposed_by_T_1d.shape}):\n{transposed_by_T_1d}")
print(f"Is my_linspace_array == transposed_by_T_1d: {np.array_equal(vaani_array, transposed_by_T_1d)}")

row_vector = vaani_array.reshape(1, -1)
transposed_row_vector = row_vector.T
print(f"\nOriginal array reshaped to a row vector (shape {row_vector.shape}):\n{row_vector}")
print(f"Transpose of row vector using .T (shape {transposed_row_vector.shape}):\n{transposed_row_vector}")
```

### Q5. Create a 2-D Array of three rows and four columns, named `ucs420_<your_name>>` with following values – 10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35. Compute the mean, median, max, min, unique elements. Reshape the array to four rows and three columns and name it as `reshaped_ucs420_<your_name>>`. Resize the array to two rows and three columns and name it as `resized_ucs420_<your_name>>`.

```python
ucs420_vaani = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])
print(f"Original 2D Array (ucs420_array):\n{ucs420_vaani}")

print(f"\nMean of the array: {np.mean(ucs420_vaani)}")
print(f"Median of the array: {np.median(ucs420_vaani)}")
print(f"Maximum element in the array: {np.max(ucs420_vaani)}")
print(f"Minimum element in the array: {np.min(ucs420_vaani)}")
print(f"Unique elements in the array: {np.unique(ucs420_vaani)}")

reshaped_ucs420_vaani = ucs420_vaani.reshape(4, 3)
print(f"\nReshaped array (4 rows, 3 columns):\n{reshaped_ucs420_vaani}")
resized_ucs420_vaani = np.resize(ucs420_vaani, (2, 3))
print(f"\nResized array (2 rows, 3 columns):\n{resized_ucs420_vaani}")
```
