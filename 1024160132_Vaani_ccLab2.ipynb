```python
#q3
import random
random.seed(1024160132)
```

```python
#i
nums = [random.randint(100,900) for x in range (100)]
print(nums)
```

```python
#ii
odd = [x for x in nums if x % 2 != 0]
print(odd)
print(len(odd))
```

```python
#iii
even = [x for x in nums if x % 2 == 0]
print(even)
print(len(even))
```

```python
#iv
def is_prime(n):
  for i in range(2, n):
    if n % i == 0:
      return False
  return True

prime = [x for x in nums if is_prime(x)]
print(prime)
print(len(prime))
```

```python
#v
most_freq = max(set(nums), key = nums.count)
print(most_freq)
```

```python
# iii
print(A.intersection(B))
```

```python
# iv
print("A - B:", A.difference(B))
print("B - A:", B.difference(A))
```

```python
# v
print(A.symmetric_difference(B))
```

```python
# vii
print(" A subset of B?", A.issubset(B))
print("B superset of A?", B.issuperset(A))
```
