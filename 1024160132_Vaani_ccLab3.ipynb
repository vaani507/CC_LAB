```python
# 6
path = "employees.csv"
df = pd.read_csv(path)
print(df)

# a
print(f"Shape: {df.shape}")

# b
print("\nINFO")
print(df.info())

# c
print("\nDescriptive Stats")
print(df.describe())

# d
print("\nFirst 5 rows:")
print(df.head())
print("\nLast 3 rows:")
print(df.tail(3))

# e
print(f"Average salary of employees: {df['Salary'].mean()}")
print(f"Total bonus: {df['Bonus'].sum()}")
print(f"Youngest employee's age: {df['Age'].min()}")
print(f"Highest performance rating: {df['Rating'].max()}")

# f
df_copy = df.copy()
df.sort_values(by='Salary', ascending=False, inplace=True)
print("\nEmployees sorted by salary in descending order:")
print(df)
df = df_copy

# g
df['Category'] = 'Average'
df.loc[df['Rating'] >= 4, 'Category'] = 'Good'
df.loc[df['Rating'] >= 4.5, 'Category'] = 'Excellent'
print(df)

# h
print(f"Missing values: \n{df.isna().sum()}")

# i
df.rename(columns={'Employee_ID':'ID'}, inplace=True)
print(df)

# j
  # i
print("More than 5 YOE")
print(df[df['Years_of_Experience'] > 5])

  # ii
print("IT Dept.")
print(df[df['Department'] == "IT"])

# k
df['Tax'] = 0.10 * df['Salary']
print(df)

# l
df.to_csv("updated_employees.csv", index=False)
```
